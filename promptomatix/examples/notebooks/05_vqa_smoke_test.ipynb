# VQA smoke test

Run one Visual Question Answering call through Promptomatix before starting prompt optimization.

For OpenAI-compatible local VLM servers, set these environment variables before running the notebook:

```bash
export VQA_MODEL_PROVIDER=local
export VQA_MODEL=Qwen/Qwen2-VL-7B-Instruct
export VQA_API_BASE=http://localhost:8000/v1
export VQA_API_KEY=EMPTY
```

For OpenAI, set `OPENAI_API_KEY` or `VQA_API_KEY` and optionally `VQA_MODEL`.

````python
from pathlib import Path
import os
import sys

def find_repo_root(start: Path) -> Path:
    for candidate in (start, *start.parents):
        if (candidate / "promptomatix" / "src" / "promptomatix").exists():
            return candidate
    raise RuntimeError("Could not find repository root from the current notebook path")

repo_root = find_repo_root(Path.cwd().resolve())
src_path = repo_root / "promptomatix" / "src"
if str(src_path) not in sys.path:
    sys.path.insert(0, str(src_path))

from promptomatix import answer_visual_question

repo_root
````

````python
from IPython.display import Image, display

image_path = repo_root / "promptomatix" / "images" / "logo.png"
question = "What text is visible in this image?"

display(Image(filename=str(image_path)))
print(f"Image: {image_path}")
print(f"Question: {question}")
````

````python
provider = os.getenv("VQA_MODEL_PROVIDER", "openai")
model = os.getenv("VQA_MODEL", "gpt-4o-mini")
api_base = os.getenv("VQA_API_BASE") or os.getenv("OPTIMIZER_API_BASE")
api_key = (
    os.getenv("VQA_API_KEY")
    or os.getenv("OPTIMIZER_API_KEY")
    or os.getenv("OPENAI_API_KEY")
)

if provider.lower() == "local" and not api_key:
    api_key = "EMPTY"

print(f"Provider: {provider}")
print(f"Model: {model}")
print(f"API base: {api_base or '(provider default)'}")
````

````python
answer = answer_visual_question(
    image=image_path,
    prompt=question,
    model_provider=provider,
    model_name=model,
    model_api_key=api_key,
    model_api_base=api_base,
    max_tokens=256,
)

print(answer)
````
